## ML Data Preparation

#### ✅ Step 1: Create an ML Dataset

#### Load your Gold table: After validation

```python
gold_df = spark.table('retail_project.gold.gold_validation_fact_sales')
```

```python
gold_df.display()
```

#### Step 2: Keep Only Rows with a Target Value

- Since we're predicting `Total_revenue`, rows without a target can't be used for supervised learning.

```python
# Filer the Dataset only contains Not Null Values in th target Column - Total Revenue
from pyspark.sql.functions import col

ml_df = gold_df.filter(col('Total_revenue').isNotNull())
```

```python
# Check the count

ml_df.count()
```

```python
display(ml_df)
```

#### Step 3: Verify Remaining Nulls

```python
from pyspark.sql.functions import col, when, count

null_df = ml_df.select([
    count(when(col(column).isNull(), column)).alias(column)
    for column in ml_df.columns
])

display(null_df)
```

After filtering out rows where **`Total_revenue` is NULL**, your missing values have reduced significantly.

### **Inference After Removing NULL Values from `Total_revenue`**

* **`Total_revenue` and all revenue-related columns** (`Avg_revenue`, `Min_price`, `Max_price`) now have **0 missing values**, making the target variable complete and suitable for analysis and machine learning.
* **Customer details** (`customer_name`, `gender`, `city`, `state`) still have **496 missing values**, indicating incomplete customer demographic information.
* **`loyalty_points`** and **`Avg_loyalty_points`** have **2,701 missing values**, suggesting many customers are not enrolled in the loyalty program or loyalty data is unavailable.
* **`product_id`** and **`customer_id`** have only **70 missing values** each, indicating that most records have valid identifiers.
* **Core transaction columns** (`product_name`, `category`, `order_date`, `Year`, `Month`, `Total_quantity`, and `Total_orders`) contain **no missing values**, indicating the transactional data is clean and reliable.

### **Key Insight**

Filtering out records with missing **`Total_revenue`** has improved data quality by ensuring the target variable is complete, leaving only a small amount of missing customer information and loyalty-related data for further handling.


#### Step 4: Save the ML Dataset

```python
ml_df.write.format('delta').mode('overwrite').saveAsTable('retail_project.gold.ml_sales_dataset')
```
